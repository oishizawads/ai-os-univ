# Colab → VSCode Remote Tunnel

Colab の GPU/ストレージをローカル VSCode から叩くためのスターター。

## 前提
- ローカル VSCode に **Remote - Tunnels** 拡張がインストール済み
- 同じ GitHub アカウントで VSCode / Colab 両方にサインインする

## 流れ
1. 下のセルを上から順に実行
2. `login()` の出力にある GitHub device code で認証
3. `connect()` の出力に出る `https://vscode.dev/tunnel/<name>` を開くか、ローカル VSCode の Command Palette → `Remote Tunnels: Connect to Tunnel...` で接続

## 注意
- このノートのタブを閉じると接続が切れる
- 無料 Colab は最大 ~12h でランタイム終了。切れたら login/connect をやり直す
